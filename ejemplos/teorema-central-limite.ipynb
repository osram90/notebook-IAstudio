# Teorema central del límite

Se toman muestras de tamaño `n` de una distribución uniforme (que no es normal) y se observa cómo la media de las muestras se acerca a una normal.

```python
import numpy as np

rng = np.random.default_rng(42)
N = 10_000   # número de muestras
n = 30       # tamaño de cada muestra

medias = rng.uniform(0, 1, size=(N, n)).mean(axis=1)
print(f'Media de las medias: {medias.mean():.4f}  (teórico 0.5)')
print(f'Desviación de las medias: {medias.std():.4f}  (teórico {np.sqrt(1/12/n):.4f})')
```

La media de las muestras se comporta como una normal con centro 0.5 y desviación `1/√(12n)`, aunque la población sea uniforme.

```python
from collections import Counter

conteo, bordes = np.histogram(medias, bins=12, range=(0.3, 0.7))
for c, b in zip(conteo, bordes):
    print(f'{b:.3f} | ' + '#' * (c // 100))
```
